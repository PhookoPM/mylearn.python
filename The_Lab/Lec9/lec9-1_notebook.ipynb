### 11-1.1.2 Three Key Points in Consumer Behavior Data Analysis

In this chapter, we will explain representative methods for analyzing consumer behavior, along with Python implementations, so that you can apply them yourself. However, it is important to note that each method is just a framework for analysis. So, what is the most important thing when conducting analysis? The answer is, as expected, having a clear analysis objective and executing it in line with the marketing strategy of the company or business. While marketing strategy is a broader topic that will be addressed later, here we will first focus on the importance of having a clear analysis objective.

While it would be ideal to have access to various types of consumer data as mentioned above, there are often cases where the desired data is not available. Additionally, it is nearly impossible to collect all data about customers. Trying to achieve this would require setting up IT infrastructure (which will be briefly discussed in a later section), and it could involve significant time and costs.

Furthermore, as we will explain later using specific POS data, even if the sales of a product are increasing, it is unlikely that the cause can be identified from the POS data alone. With limited data, the insights that can be gained are also limited. While it may seem obvious, conducting data analysis without a clear objective or blindly manipulating data without proper design makes it difficult to derive meaningful insights.


So, what should be done? Here, three key points will be discussed.  
The first point is to "have a hypothesis."  
The second point is whether the analysis leads to "actions that can trigger change through experimentation".  
The third and final point is "comparison."

1. Hypothesis-driven objectives
2. Actionable insights
3. Method comparisons

A very simple example would be considering a product recommendation for a customer segment group (let's call it X). Suppose you are unsure whether it is better to recommend product A, which has no purchase history, or product B. In this case, you might discover that other customers with similar preferences and behavior history to those in segment X have previously purchased product A. This leads to the hypothesis that recommending product A to customers in the X group might increase the likelihood of them making a purchase. This is the first key point. (However, this is just a simple example; in practice, much more complex approaches are used.)


Next, regarding the second point, actionable insights, the actual outcome of the hypothesis cannot be known until it is tried. Therefore, being able to conduct an "experiment" to test the hypothesis is crucial. This means recommending product A to the customer group X (via a website, direct contact, email, etc.). This is the concrete action. Sometimes, analysis is conducted, but it may be unclear how to apply the findings or what improvements will result. This situation occurs when there is no connection to action or causality. In the current example, the analysis of the customer group's behavior and the experiment of influencing that behavior through the recommendation links to change, which is why this is the second key point, "actionable insights."

Finally, the third important point in analysis is method comparisons. Analysis is essentially synonymous with comparison and is critical. There are various methods and dimensions for comparison, but in the example above, you could randomly select a few customers from segment X to form a group that will not receive any recommendation (this is called the control group). The group that receives the recommendation for product A is referred to as the test group. By analyzing the differences in purchase rates between the test group and the control group, you can determine whether the recommendation was effective. This is similar to an A/B test on the web, where you define comparison groups and adopt the strategies that show positive results.

Other methods of comparison may involve not just between groups, but also comparing sales to the previous year or previous month. By making such comparisons, you can analyze which strategies worked and which did not, allowing for data-driven improvements. However, it is important to note that not everything should be subjected to A/B testing. For example, if the user base is too small or if it is clear that product A is superior, running an A/B test may not yield meaningful results. Additionally, if the comparison group is not selected correctly, there is a risk of falsely attributing an effect, so caution is required.

By keeping these three points in mind during analysis, it becomes possible to run the **PDCA cycle (Plan-Do-Check-Act)**, which is often mentioned as a framework for continuous improvement: hypothesis -> experiment -> verification -> improvement. If analysis cannot be done within this cycle, it can be said that it is of limited value.

However, in this lecture, since we cannot conduct experiments or verifications, we will focus on implementing Python to analyze data with a hypothesis in mind and make comparisons, all while keeping these points in mind.

### 11-1.2.2 What is KPI?

When considering marketing strategies, **KPIs** and **KGIs** become important. Many people have likely heard of these terms. KPI stands for **Key Performance Indicators** and these are the numbers that should be tracked based on the company's strategy. Simple examples include "sales revenue" or "number of new customer acquisitions." The reason these are important is that without such numbers, it becomes difficult to assess whether the strategy is working or not, and decisions may be made based on intuition rather than data.

Additionally, KGI, which can broadly be considered as part of KPI, stands for **Key Goal Indicators**. It refers to the numerical goal to be achieved. For example, sales revenue is a KPI, but setting a target figure, such as an annual sales goal of 5B USD, would be an example of a KGI.

By determining the numbers to track, verifying whether they can be achieved, and making improvements, it becomes possible to visualize more objective metrics and run the PDCA cycle effectively.

Incidentally, it is said that sales representatives at Keyence Corporation, a well-known company in sales, have hundreds of KPIs. As mentioned in the reference books, this is an extreme example. However, the mindset of setting goals, tracking numbers, constantly considering why a goal wasn't achieved, and thinking about how to improve is very important.

On the other hand, KPIs are merely a means to an end and should be used as a tool to achieve objectives. It is important to avoid setting rigid KPIs that cannot be changed and placing excessive emphasis on them. The key is understanding why KPIs are being set and managed. If not done correctly, it can lead to heading in the wrong direction. For example, achieving sales targets doesn't mean pushing sales too aggressively. Such actions can affect future sales and reputation, so it's important to set and track KPIs carefully.


Later, you will learn how to implement Python to analyze metrics such as sales trends from POS data and churn rates (cancellation rates). Of course, depending on the data volume, these metrics can sometimes be calculated in Excel, so feel free to use the appropriate tool for the task at hand.

#### Side Note 6: What indicators should be used to compare product sales?

The analysis objective here is to consider whether to change the products offered in each country. For example, let’s focus on the product with StockCode 22423. In analysis, there is a principle called “apples to apples,” meaning comparisons should be made between like-for-like metrics. To illustrate this, we introduce the Purchase Incidence (PI) metric, which is commonly used in retail marketing analysis. PI compares the sales of a target product between stores with different customer footfalls. Strictly speaking, customer footfall is used in the denominator. However, in this case, we will treat the total number of baskets as a proxy for the number of customers and calculate accordingly using POS data from the e-commerce site.

The following shows the number of baskets for each country. As expected, the UK has the highest number of purchases.

```python
country_incnt_rank_df = online_retail_data_table.groupby(["Country"])["InvoiceNo"].nunique().sort_values(ascending=False)
country_incnt_rank_df
```

```python
country_total_rank_df = online_retail_data_table[online_retail_data_table.StockCode==22423].groupby(["Country"])["TotalPrice"].sum().sort_values(ascending=False)
country_total_rank_df
```

Next, to calculate the PI value (or a similar metric), we will merge the two sets of data from above. This utilizes the full functionality of pandas, so if you're unsure, make sure to review pandas thoroughly.

```python
merge_country_df = pd.concat([country_incnt_rank_df,country_total_rank_df]
                              , axis=1
                              , join='inner')
merge_country_df.columns = ['cnt', 'totalP']
merge_country_df
```

Since we have the purchase amount and basket count for each country from the data above, the "PI value" (or a similar metric) is calculated as follows, and then sorted for ranking. (Since the PI value is the purchase index per 1000 customers passing through the register, it is multiplied by 1000.)

```python
merge_country_df['PI'] = merge_country_df.totalP / merge_country_df.cnt * 1000
merge_country_df.PI.sort_values(ascending=False)
```

### 11-1.5.2 What is Decile Analysis?

In the previous section, we discussed segmenting customers and identifying valuable customers. However, when we talk about "valuable customers," what defines "valuable" can vary depending on the industry, company, and individual perspectives.

When we say "valuable customers," one common factor is the "amount spent." It can be said that "customers who spend a lot of money are valuable customers." But even though a customer may spend a lot, it is often unclear how much they need to spend to be classified as valuable, or at what threshold to draw the line. In the **decile analysis** introduced here, all customers are ranked in descending order of purchase amount, divided into ten equal parts, and the sales composition ratio is analyzed. For example, the top two groups might be defined as valuable customers.

The reason for conducting this analysis is that it provides a way to identify valuable customers. By identifying these customers and differentiating services from those who are not valuable, companies can encourage loyal customers to become even more devoted. It would not be beneficial to treat a customer who spends only 10 cents a year the same as a customer who spends over 100,000 dollars, as this could reduce the loyalty of high-spending customers and may even lead to a decrease in continued service usage.

Of course, in practice, customer identification and the determination of valuable customers cannot always be based solely on purchase amount, and there may be regulations that prohibit such practices in certain service areas. By adding other dimensions, such as the sales amount axis and those introduced in later sections like RFM analysis, companies can deliver messages tailored to each segment and increase the potential for offering better services.

Here, the basic procedure for segmenting is learned through decile analysis. First, we will use the data solved in the exercise above, which was the sales ranking by customer.

```python
cust_total_p_rank_df.head()
```

To divide into 10 equal parts, we will use Pandas' `qcut`. This is a review of Pandas, so if you are not familiar with it, please review pandas.

```python
q_ten_cust_price_df = pd.qcut(cust_total_p_rank_df, 10)
q_ten_cust_price_df
```

Here, let's count the number of people in each of the 10 groups and verify that they are divided equally. Use the `value_counts()` method of Pandas and sort by index in ascending order.

The interpretation of the table above is as follows: There are 434 companies in the group with total sales up to 156 pounds (it might have been treated as euros during this period, but for now, we'll assume it's in pounds). The next group consists of 434 companies with sales ranging from 156 pounds to 250 pounds, and so on, with the groups divided equally.

#### Side Note 7: Implementing actions after segmentation (Experimental Design)

After conducting this analysis, it is necessary to consider how to approach and implement actions for each segment (1 to 10). Since each group represents a different customer profile, it is common to apply different strategies for each. The same applies to other analyses such as RFM analysis.

To illustrate this, let us consider a scenario where the segmentation results are provided as a sales list.

For high-value customers, it is common to engage in frequent communication and offer valuable information. After segmenting, strategies for each segment are determined. However, when it comes to executing those strategies, it may sometimes be necessary for the sales representatives to review and make decisions individually for each case.

In practical terms, some financial institutions and marketing companies carry out practices similar to what is described in RFM analysis. After segmenting customers, they prioritize approaches and change the services or information offered to each group, followed by verification of the actions that resulted. Since the target customers often number in the hundreds of thousands or millions, system processing is usually employed.

This is known as experimental design. In practice, a test group within a given segment is tested with a certain strategy, while a control group is not, in order to assess the effectiveness of the strategy. This resembles A/B testing, where the uplift between the test and control groups is calculated to measure the effect of the strategy.

Specifically, for Segment 1, where few purchases have been made, a 50% discount campaign might be introduced. For Segment 10, where purchases are already high, a different service (such as a travel voucher, a gift, or information provision) might be offered. However, since the effectiveness of these changes is uncertain, "experiments" must be conducted. The results of these experiments must be verified before a PDCA (Plan-Do-Check-Act) cycle can be implemented.

However, there are instances where control cannot be established, or where legal or regulatory restrictions make it difficult to change services for customers. Approaches and methods for conducting rigorous testing when control cannot be maintained, as well as causal inference and exploratory techniques, will be introduced in more detail in the advanced-level content.

#### Side Note 8: Decile Analysis for Time-series
Here, the decile analysis was performed for a single period only. However, it is not always the case that the top-tier, high-value customers in segment 10 remain in segment 10 indefinitely. Additionally, there might be business intentions to move customers in segment 3 up to segment 5 through improved services or targeted actions.

To analyze this, we perform **decile time-series analysis** over two periods. The purpose of this analysis is to track the changes in sales of customers who were top-tier last year but saw a drop in sales this year, and vice versa, by observing the time-series trends of customer sales. ### 11-1.5.3 What is ABC Analysis?

So far, we have focused on identifying and segmenting key customers, but now we will shift our attention to products and introduce **ABC analysis**, also known as priority analysis. The basic approach is similar to decile analysis, but instead of focusing on customers, we analyze products. ABC analysis is used to evaluate products, for example, to identify best-sellers and slow-moving items.

In ABC analysis, products are categorized as follows: A accounts for up to 80% of cumulative share, B for up to 95%, and C for the rest. These thresholds can be adjusted. The concept is similar to the **Pareto principle**, and the analysis helps determine whether A and B products should be prioritized for handling.

As an additional note, ABC analysis is often conducted within product categories. For instance, analysis might be performed specifically on the beer category within alcoholic beverages or on a particular category of cosmetics, such as lipsticks. Narrowing down the product scope is a common practice.

Regarding category classification, it is typically predefined but not entirely fixed. Whether a product is categorized as A or B may depend on the analyst's judgment. Often, a product category master is defined before analysis. If you understand decile analysis or the Pareto principle, ABC analysis is straightforward. However, since we lack a product category master in this case, we will conduct ABC analysis on all products. When a product category master is available, it can simply be joined with the results afterward, so the implementation remains fundamentally unchanged. ### 11-1.5.4 What is RFM Analysis?

Now, let's shift the focus back to customers. In decile analysis, customers were classified based solely on their purchasing amounts. However, the question of whether customers who spend more are always better or more valuable is a separate issue.

For instance, even if two customers both spend \$10,000, the approach will differ depending on whether they made the purchase within the last month or ten years ago. Similarly, a customer who spends \$10,000 in a single purchase behaves differently from one who spends \$100 spread over 100 transactions.

Therefore, here we will introduce a method called **RFM analysis** to more appropriately segment valuable customers for your business, going beyond just total purchase amounts. The RFM analysis stands for **Recency (the most recent purchase date), Frequency (total number of purchases), and Money (total amount spent)**, and is an approach to classify customers based on these three dimensions. The more a customer spends, the more recent their purchases, and the higher their purchase frequency, the more they are considered a valuable customer, and this approach helps identify them.

In RFM analysis, thresholds are defined for each of the three dimensions. For Money, for example, customers are segmented into three groups: those with low purchase amounts, those with moderate purchases, and those with high purchases. Specifically, customers who have a total amount from 0 to 100 can be classified as Rank 1, those with a total between 101 and 1000 as Rank 2, and those above 1000 as Rank 3.

There is no absolute rule for setting these thresholds, and RFM analysis can vary depending on the industry or company, with variations such as using only RF or FM.

Here, let's assume that the data was extracted on the reference date, December 10, 2011, and proceed with the analysis.

Since the Frequency (F) and Money (M) have already been analyzed and aggregated, we will now focus on calculating the Recency (R). Specifically, we will calculate how much time has passed since the last purchase, relative to the reference date of December 10, 2011.

```python
base_date = datetime.datetime(2011,12,10)
```

### 11-1.5.5 What is Churn Analysis?

Until now, the analysis has focused on existing customers. The objective has been to segment customers and identify them in order to improve the service provided to these existing customers.

Here, I will introduce **Churn analysis**. As the name suggests, churn analysis is conducted to understand why customers may leave (stop making purchases) and to take measures to address this issue. While a certain level of **churn** is unavoidable, a growing number of customers leaving can significantly impact a company's revenue. Therefore, it is crucial to quickly detect this change, understand why customers are churning, and determine how to prevent it.

Just like identifying customers, churn analysis aims to extract customers with the goal of preventing churn. The specific implementation of churn analysis can actually be done using the same approach as the decile analysis. As a KPI, the **retention rate** (1 - churn rate) is often set and monitored, so churn analysis can also be considered as analysis of retention.

Here, there are some important points to note about customer churn analysis. For services that involve ongoing contracts, such as mobile phone or insurance services, it is easier to define "churn" based on the data. On the other hand, for consumer goods (such as beer or daily necessities), it can be difficult to determine whether a customer is continuing to purchase or not. In such cases, it is necessary to define churn by setting a certain purchase period. If there is no purchase during that period, the customer is considered to have churned. In this case, we will define churn based on a set period.

So, how do we specifically define "churn" in the data? Before looking at the implementation below, think about it for a moment.

### 11-1.5.7 Sales Factor Decomposition for Problem Exploration

Here, we will explain the sales factor decomposition for problem exploration. This is not universally fixed, but there is a certain framework that can be followed to break down sales. Specifically, sales can be decomposed as follows:

Annual sales = annual sales per company $\times$ number of companies = annual sales per product type $\times$ annual purchase frequency per company $\times$ number of companies

There are various other ways to break this down. Comparing annual data may be difficult with POS data, but this approach can be used by calculating and comparing data year over year. The goal is to explore the question, "What is the most significant factor contributing to the decrease in sales?" By comparing last year's factor decomposition with this year's, you can approach the problem from a different angle.
